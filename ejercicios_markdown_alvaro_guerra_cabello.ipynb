# Markdown (I)

### ¿Qué es?

Lenguaje de marcado que nos permite aplicar formato a nuestros textos mediante unos caracteres especiales. Este lenguaje está pensado para web, pero es muy común utilizarlo con cualquier tipo de texto, independientemente de sus destino.

Lo bueno que tiene es que se edita en **texto plano y está integrado en muchísimas herramientas**, como Jupyter Notebook o RStudio.

### Markdown vs HTML

Tenemos un viejo conocido en cuanto a programación web: HTML. Son lenguajes muy diferentes. Con HTML podemos construir un complejo **árbol de tags**, mientras que markdown se desarrolla en texto plano. Por supuesto, las finalidades son distintas. Ya veras a lo largo de este notebook, que ambos lenguakes son perfectamente compatibles. 

### ¿Cómo funciona?

Contiene una serie de **caracteres especiales** que le dan forma a los textos.

### ¿De qué nos va a servir?

En Jupyter lo normal será crear celdas con código, pero también tenemos la posibilidad de insertar cerlas de markdown, donde podrmos poner **imágenes, títulos, enumerar texto, listar, citar y mucho más**.

## 3. HTML

Markdown nos ofrece la posibilidad de escribir código HTML dentro de una celda markdown. Si temanejas bien con HTML y quieres insertar una poricón de código de este lenguaje, markdown lo va a interpretar.

<h1>Header 1</h1>
<h2>Header 2</h2>
<h3>Header 3</h3>

## 4. Negrita, cursiva

Para resaltar texto en negrita tenemos que rodearlo con asteriscos. En el caso de que queramos cursiva, será un único asterisco, y si deseamos combinar negrita con cursiva, son 3 asteriscos.

**Texto en negrita** 

*Texto en cursiva*

***Texto en negrita y cursiva***

Cuidado con dejar espacios entre los asteriscos y el texto. Es decir, si queremos escribiren negrita, inmediatamente despues de los asteriscos tiene que ir el texto: ** No es negrita **

## 5. Citar

En ocasiones, resulta útil poner una citación, o una nota, destacándola con un margen. Esto lo podemos hacer mediante el símbolo mayor que “>”. Con doble >> se consigue un segundo nivel de cita.

>Recuerda la importancia del conocimiento del negocio y del dominio
>>Y utilizar atajos de teclado

## 6. Saltos de Línea
Para poder separar párrafos y líneas en una misma celda, Markdown es un poco especial. No detecta los saltos de línea tal cual. Tenemos que indicarselo de nuevo con una "marca".

"En un lugar de la Mancha"
"Veinte años después frente al pelotón de fusilamiento"

La marca para indicarle a Markdown que debe considerar una línea nueva son dos espacios seguidos.

"En un lugar de la Mancha"

"Veinte años después frente al pelotón de fusilamiento"

## 11. Imágenes
Si tenemos una imagen en el ordenador, en general le vamos a decir a Markdown que apunte a esa imagen. Lo normal es tener todas las imagenes agrupadas en una carpeta dentro de tu repositorio.

Usamos la sintaxis ![ nombre cualqueira](ruta de la imagen).

La imagen tiene que estar en la misma carpeta que este notebook, debido a la sintaxis `./imagen.png`. Con el `./` Jupyter entiende que tiene que buscar en la carpeta de este notebook:

1. `./imagen.png` si la imagen esta en el mismo directorio donde está este Notebook
2. `./img/imagen.png` si dentro del directorio donde se encuentra este Notebook, hay una carpeta llamada "img", y dentro se encuentra la imagen.

#### Imagen en el mismo directorio que el Notebook

#### Imagen en el directorio "img"
