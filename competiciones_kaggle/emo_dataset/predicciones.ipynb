```python
import pandas as pd
import numpy as np
import os

from PIL import Image
from matplotlib.image import imread
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import OneHotEncoder

from tensorflow import keras
```

```python
test_dir = "data/images/test"

X_test = []
ids = []

for img_name in sorted(os.listdir(test_dir)):
    img_path = os.path.join(test_dir, img_name)

    img = Image.open(img_path).convert("L")
    img = img.resize((48, 48))
    img = np.array(img)

    img = img[..., np.newaxis]
    img = img / 255.0

    X_test.append(img)
    ids.append(img_name.replace(".jpg", ""))

X_test = np.array(X_test, dtype=np.float32)

X_test.shape
```

```python
modelo_cnn_ok = keras.models.load_model("modelos/model_cnn_ok.keras")
modelo_cnn_ok.summary()
```

```python
sample_sub = pd.read_csv("sample_submision.csv")
ids_sorted = sample_sub['id_img'].astype(str).tolist()
```

```python
X_test_sorted = []
for img_id in ids_sorted:
    img_path = f"data/images/test/{img_id}.jpg"
    img = Image.open(img_path).convert("L")
    img = img.resize((48, 48))
    img = np.array(img)[..., np.newaxis] / 255.0
    X_test_sorted.append(img)

X_test_sorted = np.array(X_test_sorted, dtype=np.float32)
```

```python
y_pred = modelo_cnn_ok.predict(X_test_sorted)
```

```python
y_pred.shape
```

```python
print(y_pred[0])
```

```python
y_labels = np.argmax(y_pred, axis=1)
```

```python
label_map = {0:'angry', 1:'disgust', 2:'fear', 3:'happy', 4:'neutral', 5:'sad', 6:'surprise'}
y_labels_text = [label_map[i] for i in y_labels]
```

```python
submission = pd.DataFrame({
    'id_img': ids_sorted,
    'label': y_labels_text
})
```

```python
submission.to_csv("submission.csv", index=False)
```
