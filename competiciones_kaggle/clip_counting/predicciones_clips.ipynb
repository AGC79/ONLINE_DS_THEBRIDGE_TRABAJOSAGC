```python
import pandas as pd
import numpy as np
import os

from PIL import Image
from matplotlib.image import imread
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow import keras
import tensorflow as tf
```

#### Procesamiento de las imagenes de test + test2

```python
test_dirs = ["data/test", "data/test_2/solution"]
X_test = []
ids = []

for folder in test_dirs:
    for img_name in sorted(os.listdir(folder)):
        img_path = os.path.join(folder, img_name)
        
        # Procesamiento de las imagenes para test
        img = Image.open(img_path).convert("RGB")
        img = img.resize((64, 64))
        img = np.array(img, dtype=np.float32) / 255.0
        
        X_test.append(img)
        
        # extracción de los IDs
        img_id = int(img_name.replace("clips-", "").replace(".png", ""))
        ids.append(img_id)
```

```python
X_test = np.array(X_test)
```

```python
cnn_model_ok = keras.models.load_model("modelos/cnn_model_ok.keras")
cnn_model_ok.summary()
```

```python
y_pred = cnn_model_ok.predict(X_test)
```

```python
y_pred.shape
```

```python
# lista de IDs en el orden con el que se ha de organizar las predicciones
sample_sub = pd.read_csv("sample_submision.csv")
ids_sorted = sample_sub['id'].astype(str).tolist()
```

```python
# Convierte las predicciones de una matriz de columnas a una lista plana de números 
y_pred_flat = y_pred.flatten()
```

```python
# Creación del dataframe para subir con los ids ordenados y las predicciones
submission = pd.DataFrame({
    'id': ids_sorted,
    'clip_count': y_pred_flat
})
```
