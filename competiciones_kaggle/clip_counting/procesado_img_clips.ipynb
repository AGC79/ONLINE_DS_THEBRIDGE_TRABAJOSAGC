```python
import os
import numpy as np
from PIL import Image

# Lista de fotos procesadas para Computer Vision
lista_fotos_bn = []
# Lista de fotos procesadas para CNN
lista_fotos_rgb = []

for foto in os.listdir("data/train"):
    ruta = os.path.join("data/train", foto)

    # Imagen en BN para Computer Vision
    img_bn = Image.open(ruta).convert("L")
    img_resized_bn = img_bn.resize((64, 64))
    # Convierte la imagen en un array de numpy y le añade una tercera dimensión del BN
    img_array_bn = np.array(img_resized_bn)[..., np.newaxis]
    lista_fotos_bn.append(img_array_bn)

    # Imagen en RGB para CNN
    img_rgb = Image.open(ruta).convert("RGB")
    img_resized_rgb = img_rgb.resize((64, 64))
    # Aquí la imagen ya viene compuesta por las 3 capas (3 dimensiones)
    img_array_rgb = np.array(img_resized_rgb) 
    lista_fotos_rgb.append(img_array_rgb)

print("BN:", len(lista_fotos_bn))
print("RGB:", len(lista_fotos_rgb))
```

## X e y para Computer Vision
