# APUNTES API

A la hora de hacer APIs, es muy importante castear los errores para que la API no rompa al usuario.

Las APIs sirven mucho para limitar y gestionar accesos. Las APIs son sistemas para controlar el acceso a los datos según los endpoints que intersa enseñar.

## Código para para crear una API

jsonify: permite trabajar con objetos json. La APIs trbajan con json

Las APIs trabajan con endpoints.

Endpoint: cada punto de acceso, cada llamada.

Toda API tiene un endpoint inicial.

get = es como un select en SQL
post = enviar/modificar datos y almacenarlos

Postman es como thunder client
Es una herramienta externa que usas para probar si tu código de Flask funciona bien. Envía peticiones (GET, POST, etc.) a tu servidor para ver qué responde. Es el "comensal" que pide la comida y comprueba si está buena.


En el código de un endpoint se puede meter un modelo predcitivo que devuelva algo cuando se le pasan ciertos datos como argumentos. Se puede entrenar o reentrenar un modelo, limpiar unos datos...
