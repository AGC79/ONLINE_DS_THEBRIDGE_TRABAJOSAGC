# MODELOS NO SUPERVISADOS

La característica principal es que trabaja con datos no etiquetados y carecen de target.

Datos no etiquetados como imágenes siginica que sus caracterisiticas no se conocen hasta que son etiquetados.

El problema de las imagenes es que no estan clasificadas. Se meten en distintos clusters en base a características de la imagen.

Estos modelos de preprocesamiento de datos/imagenes para pasárselos luego a modelos supervisados.

2 tipos:

    - Clustering: funciona por cercanía
    - Reducción de dimensionalidad: PCAs funcioa por contagio.
