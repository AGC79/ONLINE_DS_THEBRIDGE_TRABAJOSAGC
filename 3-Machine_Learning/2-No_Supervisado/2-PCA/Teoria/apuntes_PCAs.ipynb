# 📊 Varianza: explicación con ejemplo (edades)

La **varianza** mide qué tan dispersos están los datos respecto a la **media**.

- Si la varianza es **baja**, las edades están **parecidas**.
- Si la varianza es **alta**, las edades están **muy dispersas**.

---

## ✅ Ejemplo con edades

Usaremos estas edades:

Edades = [20, 22, 30, 28, 25]


### 1️⃣ Calcular la media

$$
\text{media} = \frac{20 + 22 + 30 + 28 + 25}{5} = 25
$$


### 2️⃣ Distancia de cada edad a la media

| Edad | Edad - Media | Distancia |
|------|--------------|-----------|
| 20   | 20 − 25 = -5 | -5        |
| 22   | 22 − 25 = -3 | -3        |
| 30   | 30 − 25 = 5  | 5         |
| 28   | 28 − 25 = 3  | 3         |
| 25   | 25 − 25 = 0  | 0         |

### 3️⃣ Elevar al cuadrado cada distancia

| Distancia | Cuadrado |
|-----------|----------|
| -5        | 25       |
| -3        | 9        |
| 5         | 25       |
| 3         | 9        |
| 0         | 0        |

### 4️⃣ Calcular la varianza

$$
\text{varianza} = \frac{25 + 9 + 25 + 9 + 0}{5} = \frac{68}{5} = 13.6
$$

---

## 📌 Interpretación

- La varianza es **13.6**, lo que indica una **dispersión moderada**.
- Las edades están relativamente cerca de la media (25), pero no son todas iguales.

---

## 🧠 En palabras simples

La **varianza** es:

> *El promedio de las distancias al cuadrado respecto a la media.*

Te dice *qué tanto se alejan las edades de la media*.
