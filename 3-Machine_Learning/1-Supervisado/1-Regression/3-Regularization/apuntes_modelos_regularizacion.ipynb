# MODELOS DE REGULARIZACIÓN

Los modelos de regularización sirven sobre todo para tratar problemas de overfitting.

Validadation error = test (ver grafico de overfitting - underfittingº)

Con los métodos de regularización podremos influir en la asignación de coeficientes o pesos de las variables de las features.

Ridge(L2), se intenta dar mas peso a las variables que menos coeficiente tienen y bajar el peso a las que tienen mas. Los modelos tienden a memorizar mucho las variables con mayor peso.

Lasso (L1). En vez de intentar regular todas las variables, se enfoca en las que considera útiles y le da coeficiente 0 a las que no considera útiles.

Estos modelos sirven cuando el modelo de regresión lineal tiene over fitting.
