# Apuntes NLP

Corpus: paquetes de texto, un parrafo, un libro entero.

Token: peso o valor/etiqueta numerica que se le va a dar a cada palabra, letra.

La maquina interpreta los valores numericos con pesos

tostadora = 15
telefono = 20
pangolin = 30

La mquina entiende que pangolin es 30 veces 1, y lo ve como el doble que tostadora

Para solucionar esto se le asignan vectores:

One hot encoding:

tostadora = [0,0,1]

telefono = [0,1,0]

pangolin = [1,0,0]

Embeddings:

STOP WORDS: palbras comunes (articulos, verbos, determinantes...) se quitan para seleccionar solo las palabras claves
