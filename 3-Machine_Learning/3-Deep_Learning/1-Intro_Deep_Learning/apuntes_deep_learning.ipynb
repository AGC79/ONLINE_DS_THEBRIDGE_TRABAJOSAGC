## Redes neuronales

Modelo que recoge a información de otros modelos y, si supera cierto umbral (accuracy), le pasa suinformación a otros. 

Es algo parecido al boosting. El boosting es lineal. 

Las redes neuronales se basan en capas. En cada capa hay un número de neuronas.

- capa de entrada: recibe los datos

- capas intermedias

- capas de salida. Para un problema de regresión es una neurona, si es un problema de clasificación, tantas neuronas como clases tenga el target

SESGO = BIAS = INTERCEPTO

Es como un regulador que evalua los pesos. Vamos a poder modificarlo.

## 📌 ¿Qué es el *bias* en un modelo de Machine Learning?

En un modelo lineal o en una neurona artificial usamos la fórmula:

\[
y = w_1x_1 + w_2x_2 + \dots + b
\]

Donde:

- **\( w_i \)** → pesos  
- **\( x_i \)** → variables de entrada  
- **\( b \)** → *bias* (también llamado **sesgo** o **intercepto**)  

---

## 🧠 ¿Qué hace realmente el *bias*?

El **bias NO evalúa ni regula los pesos**.

El **bias es un valor adicional que permite desplazar la salida del modelo**:

- En una regresión lineal → desplaza la recta hacia arriba o hacia abajo.  
- En una neurona → ajusta el *umbral de activación*.

Es decir, permite que el modelo aprenda patrones incluso cuando todas las entradas son cero.

---

## 🔧 Analogía intuitiva

Piensa en una neurona como una **puerta automática**:

- Los **pesos** son la fuerza con la que empuja cada sensor.
- El **bias** es qué tan difícil es abrir la puerta (el umbral).

Si el bias es grande → la puerta necesita más fuerza total para abrirse.  
Si es pequeño → se abre con menos fuerza.

---

## ✔️ Definición final y correcta

> **El bias (sesgo o intercepto) es un valor que se suma a la combinación lineal de las entradas y permite desplazar la salida del modelo, ajustando el umbral de decisión o activación.**

# Funciones de Activación en Redes Neuronales

Las funciones de activación transforman la salida de una neurona antes de pasarla a la siguiente capa.



## 0️⃣ Escalón (Step Function)

La función de activación de escalón es la más simple. Se usa en los perceptrones clásicos.

- **Definición:** Devuelve 0 si la entrada es negativa y 1 si la entrada es positiva (o cero).
$$
\text{Step}(x) =
\begin{cases} 
0 & \text{si } x < 0 \\
1 & \text{si } x \geq 0
\end{cases}
$$
- **Ejemplo:** Step(-3)=0, Step(2)=1
- **Uso:** Útil para decisiones binarias simples; activa la neurona solo si la entrada supera el umbral.
- **Problemática recomendada:** Clasificación binaria muy simple, lógica de decisión tipo "sí/no".

---

## 1️⃣ ReLU (Rectified Linear Unit)

ReLU es muy usada en redes profundas. Devuelve 0 si la entrada es negativa y deja pasar la entrada tal cual si es positiva.

- **Definición:** Devuelve el valor si es positivo, y 0 si es negativo.  
$$
\text{ReLU}(x) = \max(0, x)
$$
- **Ejemplo:** ReLU(-3)=0, ReLU(2)=2  
- **Uso:** Muy común en redes profundas; activa solo valores positivos.
- **Problemática recomendada:** Redes profundas, visión por computadora, detección de patrones complejos.

---

## 2️⃣ Sigmoide

La función sigmoide aplasta cualquier valor a un rango entre 0 y 1. Es útil para probabilidades.

- **Definición:** Convierte cualquier valor real en un número entre 0 y 1.  
$$
\text{Sigmoide}(x) = \frac{1}{1 + e^{-x}}
$$
- **Ejemplo:** Sigmoide(0)=0.5, Sigmoide(2)≈0.88  
- **Uso:** Bueno para salida de probabilidad en clasificación binaria.
- **Problemática recomendada:** Clasificación binaria, problemas donde se necesita interpretar la salida como probabilidad.

---

## 3️⃣ Softmax

Softmax se usa al final de redes de clasificación multiclase. Convierte un vector de valores en probabilidades que suman 1.

- **Definición:** Convierte un vector de valores en probabilidades que suman 1.  
$$
\text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
$$
- **Ejemplo:** Para [2,1,0] → Softmax ≈ [0.66, 0.24, 0.09]  
- **Uso:** Salida de redes para clasificación multiclase; asigna una probabilidad a cada clase.
- **Problemática recomendada:** Clasificación multiclase, reconocimiento de imágenes, NLP (etiquetado de categorías).


## Backpropagation

Algoritmo de entrenamiento de las redes neuronales. Tiene en cuenta los errores de las neuronas anteriores para el aprendizaje.

## Proceso entrenamiento

Parecido al de los modelos que hemos visto hasta ahora. Tambien se hace el split entre X e y, train y test.

Sin embargo, aquí coge mas peso la validación cada vez que se pasan 

## Parámetros importantes

**batch_size**: Número de muestras que se procesan antes de actualizar los pesos de la red. El concepto  
No se refiere a la validación directamente, sino al tamaño del **lote de entrenamiento** que se usa en cada iteración.  

- Cada iteración del entrenamiento usa un **batch** de datos.  
- Un valor demasiado grande puede consumir mucha memoria; uno muy pequeño puede hacer el entrenamiento más ruidoso.  
- La validación se suele hacer sobre un **conjunto separado**, no constantemente en cada batch.  

**epochs**: Número de veces que la red neuronal completa un **paso por todo el conjunto de entrenamiento**.  

- Una **epoch** significa que todos los datos de entrenamiento han pasado una vez por la red.  
- Durante cada epoch, los datos pueden dividirse en **batches**, según el `batch_size`.  
- Entrenar por más epochs puede mejorar el aprendizaje, pero demasiadas puede causar **overfitting (sobreajuste)**.


## Early Stopping

- Es una técnica para **detener el entrenamiento antes de que la red neuronal se sobreajuste**.  
- La red aprende gradualmente, ajustando los pesos en cada epoch.  
- Early stopping supervisa el rendimiento en el conjunto de **validación** y detiene el entrenamiento cuando la mejora se detiene o empeora.  
- Esto evita que la red aprenda **demasiado de los datos de entrenamiento** y pierda capacidad de generalización.


# Curvas de Error en Redes Neuronales

Cuando entrenamos una red neuronal, medimos **el error (loss)** tanto en los datos de entrenamiento (**train loss**) como en los de validación (**validation loss**) a lo largo de los epochs.

---

## 1️⃣ Curva de error de entrenamiento (Train Loss)
- Indica **qué tan bien el modelo se ajusta a los datos de entrenamiento**.
- Normalmente **disminuye** con cada epoch.
- Si baja demasiado y llega a cero, puede indicar **sobreajuste** (memoriza los datos).

## 2️⃣ Curva de error de validación (Validation Loss)
- Indica **qué tan bien el modelo generaliza a datos no vistos**.
- Inicialmente suele bajar junto con el train loss.
- Si empieza a subir mientras el train loss sigue bajando → **sobreajuste**.
- Si ambas curvas se mantienen altas → **subajuste** (underfitting).

## 3️⃣ Early Stopping
- Basado en la **curva de validación**: si no mejora durante varias epochs, el entrenamiento se detiene.
- Previene que el modelo aprenda demasiado de los datos de entrenamiento y pierda capacidad de generalización.
