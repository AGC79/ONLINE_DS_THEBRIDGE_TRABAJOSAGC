# Apuntes redes convolucionales

Las redes convolucionaes se basa en la estructura espacial, no en la disposición de los pixeles. Mas tolerancia al RGB

Convolución: operación matemática que aplica distintos métodos para transformar una imagen mediante un kernel o filtro (matriz), generando nuevas versiones de la imagen en las que se resaltan características específicas como bordes, texturas o patrones

Con estos se hacen filtros para detectar patrones como bordes, líneas, texturas o formas específicas dentro de la imagen, permitiendo que una red neuronal convolucional identifique estructuras anatómicas, resalte zonas de interés y aprenda características relevantes para tareas como clasificación, segmentación o detección de anomalías.
