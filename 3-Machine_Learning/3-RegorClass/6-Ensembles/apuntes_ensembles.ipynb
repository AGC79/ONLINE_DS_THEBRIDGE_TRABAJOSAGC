# ENSEMBLES

Permite combinar modelos trabajando simultaneamente.a

- combinacion de mismos modelos (Bagging u Boosting)

- combinacion de modelos diferentes

No podemos mezclar regresión y clasificación.

#### Clasificacion

- hard voting (valores mas repetido): se vasa en el predict, cuando interesa centrarse en la recall (ejemplo datos médicos), funciona mejor para modelos 
binarios.

- soft voting: se basa en el predict_proba

#### Rregresion

Se basa tb en el predict, saca la media de las predicciones de varios modelos
