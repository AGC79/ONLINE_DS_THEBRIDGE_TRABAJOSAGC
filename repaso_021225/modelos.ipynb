# MODELOS

```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.preprocessing import PolynomialFeatures, StandardScaler, LabelEncoder, OneHotEncoder, OrdinalEncoder
from sklearn import metrics
```

```python
df = pd.read_csv("data/limpieza.csv")
df
```

```python
df.info()
```

## Xtrain, y_train

```python
X = df[["Ram", "Weight", "Inches"]]
y = df["Price_euros"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

## Linear regression

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
pred = lr.predict(X_test)

print("MAE: ", metrics.mean_absolute_error(y_test, pred))
print("RMSE: ", metrics.root_mean_squared_error(y_test, pred))
print("MAPE: ", metrics.mean_absolute_percentage_error(y_test, pred))
print("R2: ", metrics.r2_score(y_test, pred))
```

## Linear regression con Standard Scalerº

```python
st = StandardScaler()
# Se le pasa todo el X_train sin la target, y_train tiene solo la target, recuerda
st.fit(X_train)             

# Estandariza X_train y X_test (media 0, desviación estándar 1)
X_train_st = st.transform(X_train)
X_test_st = st.transform(X_test)
```

```python
# esto hace fit y transform a la vez
# X_st = st.fit_transform(X)
```

```python
lr_st = LinearRegression()

lr_st.fit(X_train_st, y_train)
```

```python
pred_st = lr_st.predict(X_test_st)

print("MAE: ", metrics.mean_absolute_error(y_test, pred_st))
print("RMSE: ", metrics.root_mean_squared_error(y_test, pred_st))
print("MAPE: ", metrics.mean_absolute_percentage_error(y_test, pred_st))
print("R2: ", metrics.r2_score(y_test, pred_st))
```
